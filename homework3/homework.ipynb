```python
import sympy as sp
from IPython.display import Math, display
```

```python
def get_k_T(r, t):
    v = r.diff(t) # r'(t)
    a = v.diff(t) # r''(t)
    j = a.diff(t) # r'''(t)

    v_cross_a = v.cross(a)

    k = v_cross_a.norm() / (v.norm() ** 3)
    T = v_cross_a.dot(j) / (v_cross_a.norm() ** 2)

    return k, T
```

```python
# ~~~ 1.4.1 ~~~

def task_1_4_1(r, t, title):
    k, T = get_k_T(r, t)

    print(title)

    display(r.transpose())
    
    print("Кривизна k:")
    display(sp.simplify(k))
    print("Кручение T:")
    display(sp.simplify(T))

t = sp.Symbol('t', real=True)
```

> Для того чтобы линия была плоской, необходимо и достаточно, чтобы ее кручение было равно нулю. Доказать это.

$\implies \quad$ Так как кривая плоская, её вектор бинормали $\bar{\beta}$ является единичным вектором, ортогональным плоскости $\pi$, в котором лежит кривая. Тогда, по формуле Френе, 
$$
\begin{align*}
\dot{\bar{\beta}} &= -T \cdot \bar{\nu} \iff \\
0 &= -T \cdot \bar{\nu}
\end{align*}
$$

Так как кривая регулярная, $\bar{\nu} \ne 0$, откуда $T = 0$.

$\impliedby \quad$ По формуле Френе,
$$
\begin{align*}
\dot{\bar{\beta}} &= -T \cdot \bar{\nu} \iff \\
\dot{\bar{\beta}} &= 0 \cdot \bar{\nu} \iff \\
\dot{\bar{\beta}} &= 0 \iff \\
\bar{\beta} &= const
\end{align*}
$$

$\blacksquare$

    \begin{cases}

Рассмотрим отображение $$
F(x, y, z) =

\end{pmatrix}
$$

```python
x, y, z = sp.symbols('x y z')
var = [x, y, z]

F = sp.Matrix([x + sp.sinh(x) - sp.sin(y) - y, z + sp.exp(z) - x - sp.log(1+x) - 1])

display(Math("F = "), F)
```

```python
M = [0, 0, 0]
display(Math("F(M) = "), F.subs(zip(var, M)))
```

```python
DF = F.diff(x).row_join( F.diff(y) ).row_join( F.diff(z) )

display(Math("DF = "), DF)
```

```python
display(Math("DF(M) = "))

DF_M = DF.subs(zip(var, M))

DF_M
```

Так как $F(M) = (0, 0)^\top$ и минор $\det \frac{\partial F}{\partial (x, y)} = M_{xy} = -4$, то по теореме о неявной функции можно выразить эти координаты через $z$:
$$
\bar{r}(t) = (x(t), y(t), t),
$$
где $x(0) = y(0) = 0$.

Подставим координаты вектора $\bar{r}(t)$ в систему и возьмём производную по $t$:

```python
X = sp.Function('X')(t)
Y = sp.Function('Y')(t)

Ft = F.subs({x: X, y: Y, z: t})

Ft
```

```python
d_Ft_dt = Ft.diff(t)
display(Math(r"\frac{d}{dt} F(x(t), y(t), t) = "), d_Ft_dt)
```

## Задание 5

> Найти с точностью до движения пространства все кривые с постоянной ненулевой кривизной и постоянным кручением. Отдельно рассмотреть случай нулевого кручения.

### Нулевое кручение

Нулевое кручение $\iff \dot{\bar{\beta}} = 0 \iff$ кривая плоская. Тогда 
$$
\begin{cases}
\dot{\bar{\tau}} = k \cdot \bar{\nu}, \\
\dot{\bar{\nu}} = -k \cdot \bar{\tau}
\end{cases}
$$

Пусть $\bar{\tau} = (x, y)$. Тогда $\dot{\bar{\tau}} = (\dot{x}, \dot{y}) = k \bar{\nu}$. Отсюда $\dot{\bar{\nu}} = \frac{1}{k} (\ddot{x}, \ddot{y}) = -k \cdot \bar{\tau}$, то есть
$$
\begin{cases}
\ddot{x} = -k^2 x, \\
\ddot{y} = -k^2 y
\end{cases}
$$

Решаем систему ДУ, получаем ответ:
$$
\begin{cases}
x = a \cdot \sin (k t + b), \\
y = c \cdot \sin (k t + d), \\
z = const; a, b, c, d \in \mathbb{R}
\end{cases}
$$

### Общий случай

Запишем уравнения Френе:
$$
\begin{cases}
\dot{\bar{\tau}} &= k \cdot\bar{\nu}, \\
\dot{\bar{\nu}}  &= -k \cdot \bar{\tau} + T \cdot \bar{\beta}, \\
\dot{\bar{\beta}} &= -T \cdot \bar{\nu}.
\end{cases}
$$

Тогда $\ddot{\bar{\nu}} = -k \cdot \dot{\bar{\tau}} + T \cdot \dot{\bar{\beta}} = -(k^2 + T^2) \cdot \bar{\nu}$. Это система ДУ, аналогичная прошлому случаю, так что пишем ответ и идём спать (время 00:24):
$$
\begin{cases}
x = a \cdot \sin ((k^2+T^2) \cdot t + b), \\
y = c \cdot \sin ((k^2+T^2) \cdot t + d), \\
z = e \cdot \sin ((k^2+T^2) \cdot t + f); \\
a, b, c, d, e, f \in \mathbb{R}
\end{cases}
$$
