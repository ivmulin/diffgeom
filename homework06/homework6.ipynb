```python
import sympy as sp
from IPython.display import Math, display
sp.init_printing()
```

```python
def fancy_display(before_text, latex, after_text=r'', latexer=sp.latex, mather=Math):
    if mather is None:
        mather  = lambda i: i
    if latexer is None:
        latexer  = lambda i: i
    display(mather(before_text + latexer(latex) + after_text))
```

## Задание 1

> Решить задачу № 647: найти первую квадратичную форму тора
> $$x = (a + b \cos u) \cos v, \quad y = (a + b \cos u) \sin v, \quad z = b \sin u.$$


```python
u, v = sp.symbols('u,v')
a, b = sp.symbols('a,b')
```

```python
r = sp.Matrix( [(a + b * sp.cos(u)) * sp.cos(v),
               (a + b * sp.cos(u)) * sp.sin(v),
               b * sp.sin(u)]
             )

fancy_display(r'r = ', r, r'.')
```

```python
r_u = r.diff(u)
r_v = r.diff(v)

E = sp.simplify(r_u.dot(r_u))
F = sp.simplify(r_u.dot(r_v))
G = sp.simplify(r_v.dot(r_v))

for k, K in zip((r'E = ', r'F = ', r'G = '), (E, F, G)):
    fancy_display(k, K, r',' if K != G else r'.')
```

## Задание 2

> Решить задачу № 653: найти первую квадратичную форму поверхности
> 
> $$z = z(x, y).$$

```python
x, y = sp.symbols('x,y')
```

```python
z = sp.Function('z')(x, y)

r = sp.Matrix(
    [x, y, z]
)

fancy_display(r'r = ', r, r'.')
```

```python
r_x = r.diff(x)
r_y = r.diff(y)

E = sp.simplify(r_x.dot(r_x))
F = sp.simplify(r_x.dot(r_y))
G = sp.simplify(r_y.dot(r_y))

for k, K in zip((r'E = ', r'F = ', r'G = '), (E, F, G)):
    fancy_display(k, K, r';')

I_latex = r'\bigg(' + sp.latex(E) + r'\bigg) \mathrm{d}x^2 + 2 \cdot \bigg(' + sp.latex(F) + r'\bigg) \mathrm{d}x \mathrm{d}y + \bigg(' + sp.latex(G) + r'\bigg) \mathrm{d}y^2'

fancy_display(r'I = ', I_latex, r'.', latexer=None)
```

## Задание 3

> 3. Решить задачу № 685: найти угол между линиями
> 
> $$v = 2u, \quad v = -2u$$
> 
> на поверхности, первая квадратичная форма которой имеет вид
> 
> $$ds^2 = du^2 + dv^2.$$

```python
t = sp.Symbol('t')
```

```python
# Параметризация линий
h = sp.Matrix([t, 2 * t])
p = sp.Matrix([t, -2 * t])
```

```python
h1, h2 = h.diff(t)
p1, p2 = p.diff(t)

E = G = 1
F = 0

h_dot_h = E * (h1 **2) + F * (2 * h1 * h2) + G * (h2**2)
h_dot_p = E * (h1 * p1) + F * (h1 * p2 + h2 * p1) + G * (h2 * p2)
p_dot_p = E * (p1**2) + F * (2 * p1 * p2) + G * (p2**2)

cos = h_dot_p / sp.sqrt(h_dot_h * p_dot_p)

fancy_display(r'\cos \varphi = ', cos, r'.')
```
